```python
from langchain.memory import ChatMessageHistory
from langchain_core.messages import AIMessage, HumanMessage
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.runnables.history import RunnableWithMessageHistory

prompt = ChatPromptTemplate.from_messages(
    [
        (
            "system",
            "You are a customer service assistant for grocery online shop. User have a negative sentiment, you should try to understand their reasons. You have to speak in Persian language in simple sentence. Don't use special tokens. Don't forget to ask for order_id. the chat history is:",
        ),
        MessagesPlaceholder(variable_name="chat_history"),
        ("user", "{input}"),
    ]
)

chain = prompt | chat

class MyChatMessageHistory(ChatMessageHistory):
    def add_message(self, message):
        # Check if the message is an instance of AIMessage
        if isinstance(message, AIMessage):
            # Create a new AIMessage instance without response_metadata
            message = AIMessage(content=message.content)
        super().add_message(message)
    

demo_ephemeral_chat_history_for_chain = MyChatMessageHistory(memory_key="chat_history",
    max_len=50,
    return_messages=True,)

chain_with_message_history = RunnableWithMessageHistory(
    chain,
    lambda session_id: demo_ephemeral_chat_history_for_chain,
    input_messages_key="input",
    history_messages_key="chat_history",
)
```

```python
while True:
    inputs = input("You: ")
    decide_output = deside_chain.invoke(inputs)

    if decide_output == False:
        for i in chain_with_message_history.stream(
            {"input": inputs},
            {"configurable": {"session_id": "unused"}},
        ):
            print(i.content, end="")
    else:
        for i in return_chain.stream(inputs):
            print(i, end="")
```

```python
def decompose_string(input_string):
    # Split the string into lines
    lines = input_string.split('\n')
    
    # Extract the dissatisfaction reason and order id
    dissatisfaction_reason = lines[0].split(': ')[1]
    order_id = int(lines[1].split(': ')[1])
    
    return dissatisfaction_reason, order_id

# Usage
input_string = 'دلیل نارضیاتی: کیفیت پایین سیب ها\nشماره سفارش: 12'
dissatisfaction_reason, order_id = decompose_string(input_string)
print(f'{dissatisfaction_reason}, {order_id}')
```
